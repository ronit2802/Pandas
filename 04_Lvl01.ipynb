#### Part 4: Data Filtering (Boolean Indexing)

```python
#1 Filter df to show only rows where Sex == 'female'.

df[df['Sex'] == 'female']
```

```python
df.loc[df['Sex'] == 'female']
```

```python
#2 Filter df to find all passengers who were older than 60 years.

df[df['Age'] > 60]
```

```python
#3 Find all passengers who survived (Survived == 1) AND were in 1st class (Pclass == 1).

df[(df['Survived'] == 1) & (df['Pclass'] == 1)]
```

```python
#4 Find all passengers who paid a 'Fare' greater than $100 OR were younger than 18 years old.

df[(df['Fare'] > 100) | (df['Age'] < 18)]['Name']
```

```python
#5 Filter for passengers whose 'Pclass' is either 1 or 2 using the .isin() method.

df[df['Pclass'].isin([1,2])]
```

```python
#6 Select rows where the 'Cabin' column is NOT missing (i.e., not NaN) using .notna().

df[df['Cabin'].notna()]
```

```python
#2 Sort df by 'Fare' in descending order and display the top 5 highest-paying passengers.

df.sort_values('Fare',ascending=False).head(5)
```

```python
#4 Sort df by 'Fare' in descending order and return only the 'Name' and 'Fare' columns using .loc or column selection.

df.sort_values('Fare',ascending=False)[['Name','Fare']]
```

#### Part 4: Advanced Boolean Logic & String Filtering

```python
#1 Filter df for passengers who paid a 'Fare' between $20 and $50 (inclusive) using the .between() method.

df[['Name','Fare']][df['Fare'].between(20,30)]
```

```python
#2 Find all passengers whose 'Name' contains the word "Miss" using .str.contains().

df['Name'][df['Name'].str.contains('Miss')]
```

```python
#3 Filter df for passengers where 'SibSp' (siblings/spouses) AND 'Parch' (parents/children) are both equal to 0 (solo travelers).

df[(df['SibSp'] == 0) & (df['Parch'] == 0)]
```

```python
#4 Find all passengers who survived (Survived == 1) but were in 3rd class (Pclass == 3).

df[(df['Survived'] == 1) & (df['Pclass'] == 3)]
```

```python
#5 Filter rows where 'Embarked' is either 'C' or 'Q' AND the passenger paid a 'Fare' greater than $30.

df[((df['Embarked'] == 'C') | (df['Embarked'] == 'Q')) & (df['Fare'] > 30)]
```
