```python
import pandas as pd

# Load dataset for Level 2 practice
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

#### Part 1: Missing Values (isna(), fillna(), dropna())

```python
#1 How do you count the total number of missing (NaN) values in each column of df?

df.isna().sum()
```

```python
#2 Calculate the percentage of missing values for every column in df.

df.isna().sum()/len(df)*100
```

```python
#3 Drop all rows where the 'Embarked' column has a missing value (NaN).
df[df['Embarked']].dropna(axis = 0, inplace=True)
```

```python
#4 Fill all missing values in the 'Age' column with the median age of passengers.

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#5 Replace missing values in the 'Cabin' column with the string 'Unknown'.

df['Cabin'] = df['Cabin'].fillna('Unknown')
```

#### Part 2: Duplicates (duplicated(), drop_duplicates())

```python
#1 Write code to check if there are any completely duplicate rows in df.

df.duplicated().sum()
```

```python
#2 How many duplicate records exist if you only consider 'Ticket' and 'Fare' together?

df.duplicated(['Ticket','Fare']).sum()
```

```python
#3 Display all duplicate rows based on 'Ticket', keeping all occurrences (keep=False) to inspect them.

df.duplicated(['Ticket']).sum()
```

```python
df[df.duplicated(['Ticket'], keep=False)]
```

#### Part 3: Data Types (dtypes, astype())

```python
#1 Check the data type of every column in df.

df.dtypes
```

```python
#2 Convert the 'Survived' column from an integer (int64) to a boolean (bool).
df['Survived'] = df['Survived'].astype(bool)
```

```python
df['Survived'].dtype
```

```python
#3 Convert the 'Pclass' column into a categorical data type (category).

df['Pclass'] = df['Pclass'].astype('category')
```

```python
df['Pclass'].dtype
```

```python
#4 Convert the 'Fare' column into an integer data type (int64) by truncating decimals.

df['Fare'] = df['Fare'].astype(int)
```

```python
df['Fare'].dtype
```

```python
#5 Why does df['Age'].astype(int) throw an error if missing values are present, and how do you convert it using nullable integer types ('Int64')?

df['Age'] = df['Age'].astype('int64')
```
