## 🧹 Dirty Data (Data with Quality Issues)

Dirty data refers to data that contains **content-related errors**.

### Examples

- 🔁 Duplicated Data
- ❓ Missing Data
- 💥 Corrupt Data
- ❌ Inaccurate Data

---

## 📋 Messy Data (Data with Tidiness Issues)

Messy data refers to data that contains **structural issues**.

### Principles of Tidy Data

- ✅ Each variable forms a column
- ✅ Each observation forms a row
- ✅ Each observational unit forms a table


````python
patients = pd.read_csv('patients.csv')
treatments = pd.read_csv('treatments.csv')
treatments_cut = pd.read_csv('treatments_cut.csv')
adverse_reactions = pd.read_csv('adverse_reactions.csv')
````

````python
with pd.ExcelWriter('output.xlsx',engine='openpyxl') as writer:
    patients.to_excel(writer, sheet_name='Sheet1', index=False)
    treatments.to_excel(writer, sheet_name='Sheet2', index=False)
    treatments_cut.to_excel(writer, sheet_name='Sheet3', index=False)
    adverse_reactions.to_excel(writer, sheet_name='Sheet4', index=False)
````
