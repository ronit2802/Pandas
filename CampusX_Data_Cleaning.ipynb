Messy data refers to data that contains **structural issues**.

### Principles of Tidy Data

- ✅ Each variable forms a column
- ✅ Each observation forms a row
- ✅ Each observational unit forms a table

````python
import pandas as pd

# Extract the email address
patients_df["email"] = patients_df["contact"].str.extract(
    r'([A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,})'
)

# Extract the phone number
patients_df["phone_number"] = patients_df["contact"].str.extract(
    r'(\+?\d[\d\s().-]{7,}\d)'
)

# Clean extra spaces
patients_df["email"] = patients_df["email"].str.strip()
patients_df["phone_number"] = patients_df["phone_number"].str.strip()

# Check the results
patients_df[["contact", "phone_number", "email"]].head(10)
````
