```python
import os
import kagglehub
import pandas as pd

# Download latest version of the dataset
path = kagglehub.dataset_download("lava18/google-play-store-apps")
print("Path to dataset files:", path)

# List the files inside the downloaded folder to get the exact CSV filename
files = os.listdir(path)
print("Files downloaded:", files)

# Load the main CSV file into a Pandas DataFrame
csv_file_path = os.path.join(path, "googleplaystore.csv")
df = pd.read_csv(csv_file_path)

# Preview the data
df.head()
```
