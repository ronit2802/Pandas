```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("PromptCloudHQ/imdb-data")

print("Path to dataset files:", path)
```

```python
import os
import kagglehub
import pandas as pd
```

```python
files = os.listdir(path)

csv_file = [file for file in files if file.endswith('.csv')][0]

csv_file_path = os.path.join(path, csv_file)

df = pd.read_csv(csv_file_path)

df.head()
```

#### In Which Year There Was The Highest Average Voting?

```python
average_voting_by_year = df.groupby('Year')['Votes'].mean()
year_with_highest_average_voting = average_voting_by_year.idxmax()

print("Year with the highest average voting:", year_with_highest_average_voting)
print("Average voting in that year:", average_voting_by_year[year_with_highest_average_voting])
```

#### In Which Year There Was The Highest Average Revenue?

```python
avg_revenue_year = df.groupby('Year')['Revenue (Millions)'].mean()

avg_revenue_year.sort_values(ascending=False).head(3)
```

#### Find The Average Rating For Each Director

```python
avg_director_rating = df.groupby('Director')['Rating'].mean().sort_values(ascending=False)
avg_director_rating
```
