```python
# Date Range Function 

pd.date_range(start='2026/1/10',end='2026/1/20')
```

```python
# Only Need Business Days in the Date Range

pd.date_range(start='2026/1/10',end='2026/1/20',freq='B')
```

```python
# Only Need one day of the Week 

pd.date_range(start='2026/1/10',end='2026/1/20',freq='W')
```

### to_datetime Function 
### converts an existing objects to pandas timestamp/datetimeindex objects

```python
#Simple series Example 

s = pd.Series(['2026/1/1','2025/1/1','2024/1/1'])

pd.to_datetime(s).dt.day_name()
```

```python
# with errors 

s = s = pd.Series(['2026/1/1','2025/130/1','2024/1/1']) # in the scenario some one has entered an 130 month which is incorrect value 
pd.to_datetime(s,errors='coerce') # This is the solution for the above case
```
