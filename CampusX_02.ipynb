```python
import pandas as pd 
import numpy as np
```

```python
movies = pd.read_csv('imdb-top-1000.csv')
```

```python
movies.head()
```

```python
#Find the Top 3 Generes by Earnings

genre = movies.groupby('Genre')['Gross'].sum().sort_values(ascending=False)
```

```python
genre.head(3)
```

```python
# Find the Genre with Highest IMBD Ratings 

highest_IMDB_rating = movies.groupby('Genre')['IMDB_Rating'].max().sort_values(ascending=False).head(3)
```

```python
highest_IMDB_rating
```

```python
# Find the Genre with Highest Average IMDB Ratings

movies.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

```python
#Find the Highest Rated Movie in each Genre

highest_rated = movies.groupby('Genre')['IMDB_Rating'].max().sort_values(ascending=False)
```

```python
highest_rated
```

```python
#Find the Director with the most popularity

movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
# Find the Number of movies Done by Each Actors 

movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False).head(5)
```
