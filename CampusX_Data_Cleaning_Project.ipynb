#### Writing Summary of the Data

- The dataset consists of **1,028 rows** and **13 columns**.
- The data has been scraped from the **Smartprix** website and contains information about mobile phones listed on the platform.
- Each record represents a unique mobile device along with its pricing, hardware specifications, connectivity features, and purchase links.
- The dataset can be used to analyze smartphone pricing, hardware configurations, brand offerings, discount trends, and feature distributions across different mobile devices.

#### Column-wise Explanation

- **Mobile Phone:** Name of the mobile device along with its model description.
- **Price:** Current selling price of the mobile phone on the Smartprix website (in INR).
- **Discount:** Current discount available on the mobile device.
- **Connectivity:** Network and connectivity specifications supported by the device, such as 4G, 5G, Wi-Fi, Bluetooth, and supported bands.
- **Processor:** Processor or chipset used in the mobile device.
- **RAM and Storage:** RAM capacity and internal storage available on the device.
- **Battery:** Battery capacity and related battery specifications of the mobile device.
- **Display:** Display specifications including screen size, resolution, refresh rate, and display type.
- **Camera:** Front and rear camera specifications available on the device.
- **Memory Card:** Information about memory card support and expandable storage capacity.
- **Operating System:** Operating system and version available on the device.
- **Product URL:** URL of the product details page on Smartprix.
- **Buy URL:** URL that redirects users to the platform from which the mobile device can be purchased.
