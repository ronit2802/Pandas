```python
import pandas as pd

# Load dataset for Level 3 practice
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

#### Part 1: Basic Descriptive Statistics & Measures

```python
#1 Calculate the overall mean, median, standard deviation, and variance of the 'Fare' column.

df.head(1)
```

```python
print(f"Mean: {df['Fare'].mean()}")
print(f"Median: {df['Fare'].median()}")
```

```python
df['Fare'].describe()
```

```python
#2 Find the 25th, 50th, and 75th percentiles (quantiles) for the 'Age' column using .quantile().

df['Age'].describe()
```

```python
#3 Display summary statistics for all categorical (non-numeric) columns using df.describe(include='object').

df.describe(include='object')
```

```python
#4 Determine the skewness and kurtosis of the 'Fare' distribution using .skew() and .kurt().

print(f"Skew: {df['Fare'].skew()}")
print(f"Kurt: {df['Fare'].kurt()}")
```

```python
#5 Find the minimum and maximum 'Age' values using .min() and .max(), and identify the index of the oldest passenger using .idxmax().

print(f"Min Age {df['Age'].min()}")
print(f"Max Age {df['Age'].max()}")

print(f"Index of oldest passenger: {df['Age'].idxmax()}")
```

#### Part 2: Distribution & Frequency Analysis (value_counts())

```python
#1 Count the total number of passengers by passenger class ('Pclass').

df.head(1)
```

```python
df.value_counts('Pclass')
```

```python
#2Calculate the relative proportion (percentage) of male vs. female passengers using value_counts(normalize=True).

df.value_counts('Sex')*100/len(df)
```

```python
df.value_counts('Sex',normalize=True)*100
```

```python
#3 Count the number of passengers per boarding port ('Embarked'), including missing (NaN) values in the count.

df.value_counts('Embarked').count()
```
